```python
chat_history = []

def call_llm_api(prompt):

    import re
    match = re.search(r"my name is ([A-Za-z\s]+)", prompt, re.IGNORECASE)
    
    if match:
        name = match.group(1).strip()
        if "What is my name?" in prompt:
            return f"Your name is {name}."
        return f"Hello {name}! How can I help you today?"
    
    if "What is my name?" in prompt:
        return "I'm sorry, I don't know your name yet."
        
    return "I am a helpful assistant."

def chat_with_llm(user_input):
    context = ""
    for turn in chat_history:
        context += f"User: {turn['user']}\nAI: {turn['ai']}\n"
    
    full_prompt = f"{context}User: {user_input}\nAI:"
    ai_response = call_llm_api(full_prompt)
    chat_history.append({"user": user_input, "ai": ai_response})
    return ai_response

print("User: Hello, my name is ALI HAIDER.")
print("AI:  ", chat_with_llm("Hello, my name is ALI HAIDER."))

print("\nUser: What is my name?")
print("AI:  ", chat_with_llm("What is my name?"))
```
